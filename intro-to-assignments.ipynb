## Problem 3: Compute the Gradient for Linear Regression

Your task is to perform one step of Gradient Descent to update the weight $w$.

### The Gradient Descent Rule:
$$ w_{new} = w_{old} - \eta \cdot L'(w_{old}) $$
Where:
- $\eta$ (eta) is the **learning rate**.
- $L'(w)$ is the first derivative (gradient) of the loss function.

### Input:
1. `poly`: A string representing the polynomial loss function $L(w)$ (e.g., `"1w^2 - 4w^1 + 4"`).
2. `w_old`: A float representing the current weight value.
3. `eta`: A float representing the learning rate.

### Rules:
1. The input string will always be formatted with terms separated by ` + ` or ` - `.
2. Terms follow the format `aw^n`. If $n=0$, it is just the constant `a`.
3. You must parse the string, calculate the derivative $L'(w)$, and evaluate $L'(w_{old})$.
4. Return the new weight $w_{new}$ rounded to **4 decimal places**.

---

**Example 1:**
> **Input:** `poly = "1w^2 - 4w^1 + 4"`, `w_old = 5.0`, `eta = 0.1`
> **Explanation:** 
> 1. $L(w) = w^2 - 4w + 4$
> 2. $L'(w) = 2w - 4$
> 3. $L'(5.0) = 2(5.0) - 4 = 6.0$
> 4. $w_{new} = 5.0 - (0.1 \cdot 6.0) = 4.4$
> **Output:** `4.4`
